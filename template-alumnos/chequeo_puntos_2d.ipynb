```python
ret, det_corners = cv2.findChessboardCorners(rgb, (11, 7), None)
print(ret)
```

```python
print(det_corners)
```

```python
X = det_corners[:,0]
Y = det_corners[:,1]
print(X.shape)
print(Y.shape)
```

```python
plt.figure(figsize=(12,8), dpi=100)
plt.imshow(rgb)
plt.plot(X, Y,'ro')
# Etiquetar cada punto con el número de su fila/índice
for i in range(len(X)):
    # (x, y, texto) con un pequeño desfase (+2) para que el texto no tape el círculo
    plt.text(X[i] + 2, Y[i], str(i), color='white', fontsize=10, weight='bold')


plt.show()

```

```python
det_corners = det_corners[[0, 10, 11*6]]
X = det_corners[:,0]
Y = det_corners[:,1]
print(X.shape)
print(Y.shape)
```

```python
plt.figure(figsize=(12,8), dpi=100)
plt.imshow(rgb)
plt.plot(X, Y,'ro')
# Etiquetar cada punto con el número de su fila/índice
for i in range(len(X)):
    # (x, y, texto) con un pequeño desfase (+2) para que el texto no tape el círculo
    print(f'P_{i}: ({X[i]:.2f}, {Y[i]:.2f})')
    plt.text(X[i] + 2, Y[i], str(i), color='white', fontsize=10, weight='bold')


plt.show()

```
